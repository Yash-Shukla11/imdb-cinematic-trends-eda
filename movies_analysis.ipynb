Q24. What is the best genre for a debut director? (Highest avg rating for 1st time directors)


```python
df3 = df.assign(
    d = df['directors_name'].str.split(',') ,
    g = df['genre'].str.split(',')
).explode('d').explode('g')

df3['d'] = df3['d'].str.strip()
df3['g'] = df3['g'].str.strip()

debut_dirs = df3['d'].value_counts()
debut_dirs = debut_dirs[debut_dirs == 1].index
df3[df3['d'].isin(debut_dirs)].groupby('g')['rating'].mean()\
.sort_values(ascending= False).head()
```

Q25. Do movies with more actors (ensemble cast) earn more?

```python
df['stars_count'] = df['stars_name'].str.split(",").apply(len)
df[['stars_count' , 'gross_income']].dropna().corr()
```

Q26. Which Director has the highest "Hit Rate"? (Percentage of movies rated > 7.5)


```python
df4 = df.assign(
    d = df['directors_name'].str.split(',')
).explode('d')
df4['d'] = df4['d'].str.strip()

total = df4['d'].value_counts()

hits = df4[df4['rating']> 7.5].groupby('d').size()

hit_rate = (hits/total)*100
hit_rate.sort_values(ascending= False).head()
```

Q27. Analyse the "Sequel Effect" — Do movies with numbers in titles earn more?

```python
df['has_number'] = df['name'].str.contains(r'\d', na = False)
df.groupby('has_number')['gross_income'].agg(['mean' , 'count'])
```

 Q28. The Ultimate Business Decision: What movie should we fund?

```python
df.groupby('genre').agg(
    avg_income=('gross_income','mean'),
    movie_count=('gross_income','count')
).sort_values(by='avg_income', ascending=False).head()
```

# DATA VISUALISATION

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```


 Q1. Plot the average movie rating over the years (Line Chart)



```python
r_oy = df.groupby('year')['rating'].mean().sort_index()
```

```python
plt.figure(figsize = (10,6))
plt.plot(r_oy.index , r_oy.values , color = "darkred" , marker = 'o',markersize = 5 , linestyle = "dotted")
plt.xlabel("Year")
plt.ylabel("Average Rating")
plt.title("Average Movie Rating Over the Years")
plt.grid(alpha = 0.2)
plt.show()
```

Q2. Plot the number of movies released per year (Line Chart)



```python
movies_per_year = df['year'].value_counts().sort_index()
```

```python
plt.figure(figsize=(12,5))
plt.plot(movies_per_year.index, movies_per_year.values)

plt.title('Movies Released Per Year')
plt.xlabel('Year')
plt.ylabel('Number of Movies')

plt.show()
```

Q3. Plot the distribution of movie ratings (Histogram using Matplotlib)

Q7. Plot the distribution of Gross Income and Votes (Box Plot)

Q8. Plot the count of movies for the top actors (Count Plot using Seaborn)



```python
top10 = df['stars_name'].str.split(',').explode().str.strip() \
          .value_counts().head(10).index

df2 = df.assign(actor=df['stars_name'].str.split(',')) \
        .explode('actor')

df2['actor'] = df2['actor'].str.strip()

plt.figure(figsize=(12,5))
sns.countplot(data=df2[df2['actor'].isin(top10)],
              y='actor',
              order=top10)

plt.title('Movie Count of Top Actors')

plt.show()
```

Q9. Plot the correlation between Rating, Votes, and Gross Income (Heatmap)

```python
numeric_df = df.select_dtypes(include='number')
corr = numeric_df.corr()
plt.figure(figsize=(14,8))
sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',# It will format numbers to two decimal places
    cmap='coolwarm',
    linewidths=0.5
)
plt.title('Correlation Heatmap of Numeric Features', fontsize=18)
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.show()
```
